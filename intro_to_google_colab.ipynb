## **What is Google Colab?**

**Google Colab** is a cloud-based environment for creating and running computational notebooks in your web browser. You can write Python, run code, work with files and data, create visualizations, and share your work without installing Python or configuring a development environment.


A Colab notebook is stored separately from the computer where its code runs. Your notebook can be saved to Google Drive, while the code itself runs in a temporary **runtime** provided by Colab.

**Think of Colab as 2 things:**

1. **Notebook** = your code, text, outputs, and organization
2. **Runtime** = the computer that actually runs your code

> **📒 Note:** Your runtime can disappear or restart. Files and packages created inside the runtime are not automatically part of the notebook.

## **Working With Data**

Notebooks are one of the most common ways to work with data and Colab adds tools that make it easier.


### **Interactive Outputs**

Python libraries can produce interactive outputs directly inside a notebook.

````python
import altair as alt
import ipywidgets as widgets
from vega_datasets import data

stocks = ['AAPL', 'GOOG', 'AMZN']
df = data.stocks().query('symbol in @stocks and date > "2008-01-01"')

picker = widgets.SelectMultiple(options=stocks, value=['AAPL', 'GOOG'], description='Stock:')

@widgets.interact(symbols=picker)
def draw(symbols):
    sub = df[df.symbol.isin(symbols)]
    return alt.Chart(sub).mark_line().encode(
        x='date:T', y='price:Q', color='symbol:N'
    ).properties(width=300, height=150)
````

#### Forms

## **Runtime & Resources**

You can monitor CPU, memory, and GPU usage from the resource indicator in the top right corner of Colab.

## **Sharing & Collaboration**


### **Working With Others**

Share Colab notebooks like Google Docs with the **👥 Share** button in the top right corner of Colab.

Sharing includes the notebook's **code, text, outputs, and comments**. It does **not** include your runtime, installed packages, or temporary files.


### **Edit History**

Use revision history to see how a notebook has changed over time and restore earlier versions.

Open it from **File** > **Revision history**.

## **Exporting**

Colab notebooks use the standard `.ipynb` **Jupyter Notebook** format, so you can download them and use them outside Colab.

Open it from **File** > **Download**.

> **📒 Note:** Before sharing a notebook, decide whether to keep its outputs. They increase file size and may contain sensitive information.

To remove them, use **Edit** > **Clear all outputs**.
